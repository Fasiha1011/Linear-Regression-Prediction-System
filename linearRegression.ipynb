```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df=pd.read_csv(r"C:\Users\syeda\OneDrive\Desktop\linearregresssion\student_exam_scores.csv.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
X1 = df[['hours_studied']]
y1 = df['exam_score']
```

```python
X1.head()
```

```python
y1.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)
```

```python
model1 = LinearRegression()
model1.fit(X_train, y_train)
```

```python
y_pred = model1.predict(X_test)
```

```python
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python

import gradio as gr
import pandas as pd

def predict_exam_score(hours_studied):
    if hours_studied is None:
        return "Please enter hours studied."

    prediction = model1.predict(
        pd.DataFrame({"hours_studied": [hours_studied]})
    )[0]

    return round(float(prediction), 2)

def predict_house_value(median_income):
    if median_income is None:
        return "Please enter median income."

    prediction = model2.predict(
        pd.DataFrame({"median_income": [median_income]})
    )[0]

    return round(float(prediction), 2)

with gr.Blocks() as demo:
    gr.Markdown("# Linear Regression Prediction System")
    gr.Markdown("Predict exam scores and house values using two trained models.")

    # First dataset interface
    with gr.Tab("Student Exam Score"):
        gr.Markdown("### Predict exam score using study hours")

        hours = gr.Number(
            label="Hours Studied",
            value=5
        )

        exam_button = gr.Button("Predict Exam Score")
        exam_output = gr.Number(label="Predicted Exam Score")

        exam_button.click(
            fn=predict_exam_score,
            inputs=hours,
            outputs=exam_output
        )
    with gr.Tab("House Value"):
        gr.Markdown("### Predict house value using median income")

        income = gr.Number(
            label="Median Income (in tens of thousands)",
            value=5
        )

        house_button = gr.Button("Predict House Value")
        house_output = gr.Number(label="Predicted House Value")

        house_button.click(
            fn=predict_house_value,
            inputs=income,
            outputs=house_output
        )

demo.launch()
```

```python

from flask import Flask, request, render_template_string

app = Flask(__name__)

html = """
<h1>Linear Regression Prediction System</h1>

<h2>Student Exam Score</h2>
<form method="post">
    <input type="hidden" name="model" value="student">
    <label>Hours Studied:</label>
    <input type="number" name="value" step="any" required>
    <button type="submit">Predict Exam Score</button>
</form>

<hr>

<h2>House Value</h2>
<form method="post">
    <input type="hidden" name="model" value="house">
    <label>Median Income:</label>
    <input type="number" name="value" step="any" required>
    <button type="submit">Predict House Value</button>
</form>

<h3>{{ result }}</h3>
"""

@app.route("/", methods=["GET", "POST"])
def home():
    result = ""

    if request.method == "POST":
        value = float(request.form["value"])
        selected_model = request.form["model"]

        if selected_model == "student":
            prediction = model1.predict(
                pd.DataFrame({"hours_studied": [value]})
            )[0]
            result = f"Predicted Exam Score: {prediction:.2f}"
        else:
            prediction = model2.predict(
                pd.DataFrame({"median_income": [value]})
            )[0]
            result = f"Predicted House Value: {prediction:.2f}"

    return render_template_string(html, result=result)

print("Flask application is ready!")
```

```python

app.run(host="127.0.0.1", port=5000, debug=False, use_reloader=False)
```
